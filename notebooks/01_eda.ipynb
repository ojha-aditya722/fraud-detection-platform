## EDA Findings

- Dataset contains 284,807 transactions and 31 columns.
- Fraud is extremely rare: 492 transactions, about 0.17%.
- No missing values were found.
- Transaction amounts are strongly right-skewed.
- Fraudulent transactions do not simply correspond to larger amounts.
- Duplicate observations exist and should be handled carefully.
- Accuracy will not be a suitable primary evaluation metric because of class imbalance.
